```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Creating the DataFrame
data = {
    'Car_Model': ['Model A', 'Model B', 'Model C', 'Model D', 'Model E'],
    'Customer_Satisfaction': [4.5, 3.8, 4.2, 3.5, 4.7],
    'Sales': [150, 200, 300, 250, 180],
    'Category': ['Sedan', 'SUV', 'Sedan', 'SUV', 'Sedan']
}
```

```python

df = pd.DataFrame(data)

print(df.head())
```

```python
# # 3. Box Plot: Customer Satisfaction by Car Category

plt.figure(figsize=(12, 8))

# Box and strip plot
sns.boxplot(data=df, x='Category', y='Customer_Satisfaction', hue  = "Category", palette='coolwarm', width=0.5)
```

```python
# #
# violin_plot

plt.figure(figsize=(12, 8))
sns.violinplot(data=df,x='Category', y='Customer_Satisfaction', hue='Category',palette='viridis', inner='quart', linewidth=1.2
)
plt.title('Customer Satisfaction Distribution by Car Category', fontsize=18, fontweight='bold')
plt.xlabel('Car Category', fontsize=14)
plt.ylabel('Customer Satisfaction (Rating)', fontsize=14)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python

# #
# # 5. Heatmap: Correlation Matrix of Customer Satisfaction and Sales
plt.figure(figsize=(8, 6))
correlation_matrix = df[['Customer_Satisfaction', 'Sales']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=2, cbar_kws={'label': 'Correlation'})
plt.title('Correlation Heatmap of Customer Satisfaction and Sales', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()


```

```python

# # 7. Line Plot: Customer Satisfaction per Car Model (Line Plot)
plt.figure(figsize=(12, 8))
sns.lineplot(data=df, x='Car_Model', y='Customer_Satisfaction', marker='*', palette='Set2', linewidth=2, markersize=20)
plt.title('Customer Satisfaction per Car Model (Line Plot)', fontsize=16, fontweight='bold')
plt.xlabel('Car Model', fontsize=14)
plt.ylabel('Customer Satisfaction (Rating)', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# # 2. Bar Plot: Customer Satisfaction per Car Model
plt.figure(figsize=(12, 8))
sns.barplot(data=df, x='Car_Model', y='Customer_Satisfaction', palette='Blues_d')
plt.title('Customer Satisfaction per Car Model', fontsize=16, fontweight='bold')
plt.xlabel('Car Model', fontsize=14)
plt.ylabel('Customer Satisfaction (Rating)', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.tight_layout()
plt.show()
```

```python

# 8. Count Plot: Count of Car Categories (Sedan vs SUV)
plt.figure(figsize=(10, 8))
sns.countplot(data=df, x='Category', palette='crest', edgecolor='black', linewidth=1.5)
plt.title('Count of Car Categories', fontsize=16, fontweight='bold')
plt.xlabel('Car Category', fontsize=14)
plt.ylabel('Count', fontsize=14)
plt.tight_layout()
plt.show()
```

```python





# # Advanced Styling with Seaborn
# sns.set(style="whitegrid", palette="muted", font_scale=1.2)

# # 1. Scatter Plot: Sales vs Customer Satisfaction
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='Sales', y='Customer_Satisfaction', hue='Category', style='Category', s=100, palette='deep', markers=["o", "s"])
plt.title('Sales vs Customer Satisfaction', fontsize=16, fontweight='bold')
plt.xlabel('Sales (in units)', fontsize=14)
plt.ylabel('Customer Satisfaction (Rating)', fontsize=14)
plt.legend(title='Car Category', loc='upper left', fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
# # #
# # 2. Bar Plot: Customer Satisfaction per Car Model
plt.figure(figsize=(12, 8))
sns.barplot(data=df, x='Car_Model', y='Customer_Satisfaction', palette='Blues_d')
plt.title('Customer Satisfaction per Car Model', fontsize=16, fontweight='bold')
plt.xlabel('Car Model', fontsize=14)
plt.ylabel('Customer Satisfaction (Rating)', fontsize=14)
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.tight_layout()
plt.show()
# # #





```
